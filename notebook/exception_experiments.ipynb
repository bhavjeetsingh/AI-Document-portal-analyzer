```python
import traceback
import sys
class MyDocumentPortalException(Exception):
    """Custom exception for MyDocumentPortal."""
    def __init__(self,error_message:str,error_details:sys):
        _,_,exc_tb=error_details.exc_info()
        self.file_name=exc_tb.tb_frame.f_code.co_filename
        self.lineno=exc_tb.tb_lineno
        self.error_message=str(error_message)
        self.traceback_str= "".join(traceback.format_exception(*error_details.exc_info()))
    def __str__(self):
        return f"""
        Error in [{self.file_name}] at line [{self.lineno}]
        Message: {self.error_message}
        Traceback: {self.traceback_str}
        """
if __name__ == "__main__":
    try:
        # simulate an error
        a = 1 / 0
        print(a)  # This will raise a NameError
    except Exception as e:
        app_exc= MyDocumentPortalException(e,sys) ## i have named it MyDocumentPortalException
        logger.error(app_exc)
        raise app_exc
```

1. log this error(file, cloud config, command prompt) (it will keep the record of the error and will not stop the execution of the code)
2. raise the exception(it will stop the execution of the code)
